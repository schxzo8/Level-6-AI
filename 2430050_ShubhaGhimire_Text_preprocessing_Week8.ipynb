Removing Unwanted Text

Removing URLS.
In the below url_pattern,

https?:// → matches http or https

\S+ → matches non-space characters

| → OR

www.\S+ → matches URLs starting with www

```python
import re
def remove_urls(text):
  """
  This function will try to remove URL present in out dataset and replace it with space using regex library.
  Input Args:
  text: strings of text that may contain URLs.
  Output Args:
  text: URLs replaces with text
  """
  url_pattern = re.compile(r'https?://\S+|www\.\S+')
  return url_pattern.sub(r'',text)
```

Removing Unwanted Characters
This can be punctuation, numbers, emoji, dates and others.


```python
def remove_emoji(string):
  """
  This function will replace the emoji in string with whitespace
  """
  emoji_pattern = re.compile("["
                           u"\U0001F600-\U0001F64F"  # emoticons
                           u"\U0001F300-\U0001F5FF"  # symbols & pictographs
                           u"\U0001F680-\U0001F6FF"  # transport & map symbols
                           u"\U0001F1E0-\U0001F1FF"  # flags (iOS)
                           u"\U00002702-\U000027B0"
                           u"\U000024C2-\U0001F251"
                           "]+", flags=re.UNICODE)
  return emoji_pattern.sub(r' ', string)
```

```python
test_string = "Hi, Monika 👋🏾 , still on for the stay??? "
no_emoji = remove_emoji(test_string)
no_emoji
```

```python
#removing everyunwanted characters
def removeunwanted_characters(document):
  """
  This function will remove all the unwanted characters from the input dataset.
  Input Args:
  documet: A text data to be cleaned.
  Return:
  A cleaned document.
  """
  # remove user mentions
  document = re.sub("@[A-Za-z0-9_]+"," ", document)
  # remove hashtags
  document = re.sub("#[A-Za-z0-9_]+","", document)
  # remove punctuation
  document = re.sub("[^0-9A-Za-z ]", "" , document)
  #remove emojis
  document = remove_emoji(document)
  # remove double spaces
  document = document.replace('  ',"")
  return document.strip()       # Removes extra spaces from the start and end of a string.
```

Removeing Punctuations:

```python
from nltk.tokenize import RegexpTokenizer

def remove_punct(text):
  """
  This function removes the punctutations present in our text data.
  Input Args:
  text: text data.
  Returns:
  text: cleaned text.
  """
  tokenizer = RegexpTokenizer(r"\w+")     #\w = any word character (letters, numbers, underscore) and   + = one or more of them togethe

  lst=tokenizer.tokenize(' '.join(text))
  return lst
```

### Remove StopWord:

A majority of the words in a given text are connecting parts of a sentence rather than showing subjects, objects or intent. Word like “the” or “and” cab be removed by comparing text to a list of stopword provided by the NLTK library.

We can also define stopwords as required by our task and dataset requirement.



```python
nltk.download('stopwords')
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
stop_words = set(stopwords.words('english'))
custom_stopwords = ['@', 'RT']
stop_words.update(custom_stopwords)
```

```python

def remove_stopwords(text_tokens):
  """
  This function removes all the stopwords present in out text tokens.
  Input Args:
  text_tokens: tokenize input of our datasets.
  Returns:
  result_tokens: list of token without stopword.
  """

  result_tokens = []
  for token in text_tokens:
    if token not in stop_words:
       result_tokens.append(token)
  return result_tokens
```

## Text Normalization:

This is the idea of reducing number of words present in Corpus by the process of Lemmatization, Stemming, Capital to Lower [i.e. My -- my].
###Lemmatization:

It is an common NLP techniques used to reduce number of tokens(words) in dataset, this is acheived by replacing the word with its root words.


```python
from nltk.stem import WordNetLemmatizer
from nltk import word_tokenize,pos_tag
nltk.download('averaged_perceptron_tagger')
nltk.download('wordnet')

def lemmatization(token_text):
  """
  This function performs the lemmatization operations as explained above.
  Input Args:
  token_text: list of tokens.
  Returns:
  lemmatized_tokens: list of lemmatized tokens.
  """
  lemma_tokens = []
  wordnet = WordNetLemmatizer()
  lemmatized_tokens = [wordnet.lemmatize(token, pos = 'v') for token in token_text]

  return lemmatized_tokens


```

### Stemming:

Also a token(word) reduction techniques. This techniques tries to reduce by chopping off a part of the word at the tail end.


```python
from nltk.stem import PorterStemmer

def stemming(text):
  """
  This function performs stemming operations.
  Input Args:
  token_text: list of tokenize text.
  Returns:
  stemm_tokes: list of stemmed tokens.
  """
  porter = PorterStemmer()
  stemm_tokens = []
  for word in text:
    stemm_tokens.append(porter.stem(word))
  return stemm_tokens
```

```python
def lower_order(text):
  """
  This function converts all the text in input text to lower order.
  Input Args:
  token_text : input text.
  Returns:
  small_order_text : text converted to small/lower order.
  """
  small_order_text = text.lower()
  return small_order_text

# Test:
sample_text = "This Is some Normalized TEXT"
sample_small = lower_order(sample_text)
print(sample_small)
```

```python
def text_cleaning_pipeline(text, rule="lemmatize"):
    """
    Cleans and preprocesses a single string of text.
    """
    # 1. Basic Cleaning
    # Note: We perform these on the string directly
    text = lower_order(text)
    text = remove_urls(text)
    text = remove_emoji(text)
    text = removeunwanted_characters(text)

    # 2. Tokenization
    tokens = text.split()

    # 3. Stopword Removal
    tokens = [word for word in tokens if word not in stop_words]

    # 4. Normalization (Lemmatization or Stemming)
    if rule == "lemmatize":
        wordnet = WordNetLemmatizer()
        tokens = [wordnet.lemmatize(word, pos='v') for word in tokens]
    elif rule == "stem":
        porter = PorterStemmer()
        tokens = [porter.stem(word) for word in tokens]
    else:
        # Default to no normalization if rule is unrecognized
        pass

    return " ".join(tokens)
```

Feature Consolidation and Noise Reduction
In the Trump tweets dataset, the original text contains significant "noise" such as inconsistent casing (e.g., uppercase and lowercase), URLs, emojis, and special characters. Without preprocessing, a machine learning model treats each variation of a word as a unique token, which dilutes the statistical importance of the actual message. By converting text to a uniform case and removing elements like URLs and emojis, these variations are consolidated into single features. This reduction in noise ensures the model focuses on core content rather than stylistic inconsistencies, leading to more stable and accurate analysis.

Information Density through Tokenization and Stopword Removal
Raw tweets are often filled with high-frequency "filler" words such as "the," "is," and "of" which act as linguistic glue but carry no specific sentiment or topical value. By tokenizing the text and filtering out these stopwords, the information density of the dataset is increased. This allows meaningful terms unique to the tweets to surface as primary drivers of the analysis, making it significantly easier for the model to identify specific themes and patterns within the political discourse.

Semantic Mapping via Lemmatization
The implementation of lemmatization is vital for standardizing vocabulary by mapping different morphological forms of a word back to their base dictionary root, or lemma. For example, words like "voting," "voted," and "votes" all refer to the same fundamental concept. Without lemmatization, a model would analyze these as separate data points, but reducing them to the lemma "vote" ensures the system understands the underlying semantic intent. This process shrinks the total vocabulary size, reducing the computational burden while allowing the model to generalize much better across the entire dataset.
